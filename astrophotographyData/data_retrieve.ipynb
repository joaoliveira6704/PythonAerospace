```python
imgList = ["2MASS_h.fits", "2MASS_k.fits", "2MASS_j.fits"]
```

```python
imgDataList = []
```

```python
for file in imgList:
    with fits.open(file) as hdul:
        image_data = hdul[0].data
        imgDataList.append(image_data)
```

```python
image_stack = np.array(imgDataList)
```

```python
stacked_image = np.median(image_stack, axis=0)
```

```python
plt.imshow(stacked_image, cmap="inferno")
```
